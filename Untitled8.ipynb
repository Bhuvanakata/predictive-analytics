```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np # Import numpy for np.nan

# --- 1. Load the Data ---
# Replace 'loan_data.csv' with the actual path to your dataset.
# Make sure your CSV has columns like 'Gender', 'Married', 'Education', 'Self_Employed',
# 'ApplicantIncome', 'CoapplicantIncome', 'LoanAmount', 'Credit_History', 'Property_Area', 'Loan_Status'
# 'Loan_Status' should be your target variable (e.g., 'Y' for approved, 'N' for rejected)
loan_status_missing = False
try:
    df = pd.read_csv('loan data.csv')
    print("Dataset loaded successfully.")
except FileNotFoundError:
    print("Error: 'loan data.csv' not found. Please make sure the file is in the correct directory.")
    print("Creating a dummy dataset for demonstration purposes.")
    # Create a dummy DataFrame if the file isn't found for demonstration
    data = {
        'Loan_ID': ['LP001002', 'LP001003', 'LP001005', 'LP001006', 'LP001008', 'LP001011', 'LP001013', 'LP001014', 'LP001018', 'LP001020'],
        'Gender': ['Male', 'Male', 'Male', 'Male', 'Male', 'Male', 'Male', 'Male', 'Male', 'Male'],
        'Married': ['No', 'Yes', 'Yes', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'Yes', 'Yes'],
        'Dependents': ['0', '1', '0', '0', '0', '2', '0', '2', '2', '0'],
        'Education': ['Graduate', 'Graduate', 'Graduate', 'Not Graduate', 'Graduate', 'Graduate', 'Not Graduate', 'Graduate', 'Graduate', 'Graduate'],
        'Self_Employed': ['No', 'No', 'Yes', 'No', 'No', 'Yes', 'No', 'No', 'No', 'No'],
        'ApplicantIncome': [5849, 4583, 3000, 2583, 6000, 5417, 2333, 3036, 4006, 3200],
        'CoapplicantIncome': [0, 1508, 0, 2358, 0, 4196, 1516, 2504, 1526, 700],
        'LoanAmount': [np.nan, 128, 66, 120, 141, 267, 95, 158, 168, 70],
        'Loan_Amount_Term': [360, 360, 360, 360, 360, 360, 360, 360, 360, 360],
        'Credit_History': [1, 1, 1, 1, 1, 1, 1, 0, 1, 1],
        'Property_Area': ['Urban', 'Rural', 'Urban', 'Urban', 'Urban', 'Urban', 'Urban', 'Semiurban', 'Urban', 'Urban'],
        'Loan_Status': ['Y', 'N', 'Y', 'Y', 'Y', 'Y', 'Y', 'N', 'Y', 'Y'] # Added Loan_Status column
    }
    df = pd.DataFrame(data)
    loan_status_missing = True # Set flag since we used dummy data

# Explicitly add a placeholder 'Loan_Status' column if it doesn't exist
if 'Loan_Status' not in df.columns:
    print("\nWarning: 'Loan_Status' column not found in the dataset.")
    print("Adding a placeholder 'Loan_Status' column for descriptive analysis.")
    df['Loan_Status'] = 'Unknown' # Add a placeholder column
    loan_status_missing = True

print("\n--- Data Head ---")
print(df.head())

print("\n--- Data Information ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

# Simple Imputation for demonstration (for descriptive analytics, sometimes basic imputation is enough)
# For numerical columns, fill with median
for col in ['LoanAmount', 'ApplicantIncome', 'CoapplicantIncome']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

# For categorical columns, fill with mode
for col in ['Gender', 'Married', 'Dependents', 'Self_Employed', 'Credit_History']:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mode()[0])

# Impute missing values for Loan_Amount_Term
if 'Loan_Amount_Term' in df.columns:
    df['Loan_Amount_Term'] = df['Loan_Amount_Term'].fillna(df['Loan_Amount_Term'].mode()[0])


print("\n--- Missing Values After Imputation ---")
print(df.isnull().sum())


# --- 3. Descriptive Analytics ---

print("\n-- Distribution of Loan Status ---")
if 'Loan_Status' in df.columns and not loan_status_missing: # Only analyze if Loan_Status was in original data
    loan_status_counts = df['Loan_Status'].value_counts()
    print(loan_status_counts)
    plt.figure(figsize=(6, 4))
    sns.countplot(x='Loan_Status', data=df, palette='viridis')
    plt.title('Distribution of Loan Status')
    plt.xlabel('Loan Status (Y: Approved, N: Rejected)')
    plt.ylabel('Number of Applicants')
    plt.show()
elif loan_status_missing:
    print("Cannot show distribution of Loan Status as the column was missing or placeholder data was used.")


print("\n--- Loan Status by Categorical Features ---")

# Gender vs. Loan Status
if 'Gender' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    print("\nGender vs. Loan Status:")
    gender_loan = pd.crosstab(df['Gender'], df['Loan_Status'])
    print(gender_loan)
    gender_loan.div(gender_loan.sum(1).astype(float), axis=0).plot(kind="bar", stacked=True, figsize=(6, 4))
    plt.title('Loan Approval Rate by Gender')
    plt.ylabel('Proportion')
    plt.show()


# Married vs. Loan Status
if 'Married' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    print("\nMarried vs. Loan Status:")
    married_loan = pd.crosstab(df['Married'], df['Loan_Status'])
    print(married_loan)
    married_loan.div(married_loan.sum(1).astype(float), axis=0).plot(kind="bar", stacked=True, figsize=(6, 4))
    plt.title('Loan Approval Rate by Marital Status')
    plt.ylabel('Proportion')
    plt.show()

# Education vs. Loan Status
if 'Education' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    print("\nEducation vs. Loan Status:")
    education_loan = pd.crosstab(df['Education'], df['Loan_Status'])
    print(education_loan)
    education_loan.div(education_loan.sum(1).astype(float), axis=0).plot(kind="bar", stacked=True, figsize=(7, 5))
    plt.title('Loan Approval Rate by Education')
    plt.ylabel('Proportion')
    plt.show()

# Credit History vs. Loan Status
if 'Credit_History' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    print("\nCredit History vs. Loan Status:")
    credit_loan = pd.crosstab(df['Credit_History'], df['Loan_Status'])
    print(credit_loan)
    credit_loan.div(credit_loan.sum(1).astype(float), axis=0).plot(kind="bar", stacked=True, figsize=(6, 4))
    plt.title('Loan Approval Rate by Credit History (1: Good, 0: Bad)')
    plt.ylabel('Proportion')
    plt.show()

# Property Area vs. Loan Status
if 'Property_Area' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    print("\nProperty Area vs. Loan Status:")
    property_loan = pd.crosstab(df['Property_Area'], df['Loan_Status'])
    print(property_loan)
    property_loan.div(property_loan.sum(1).astype(float), axis=0).plot(kind="bar", stacked=True, figsize=(7, 5))
    plt.title('Loan Approval Rate by Property Area')
    plt.ylabel('Proportion')
    plt.show()

print("\n--- Loan Status by Numerical Features (Distributions) ---")

# Applicant Income vs. Loan Status
if 'ApplicantIncome' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Loan_Status', y='ApplicantIncome', data=df, palette='pastel')
    plt.title('Applicant Income Distribution by Loan Status')
    plt.xlabel('Loan Status')
    plt.ylabel('Applicant Income')
    plt.ylim(0, df['ApplicantIncome'].quantile(0.99)) # Limit y-axis to avoid extreme outliers
    plt.show()

# Loan Amount vs. Loan Status
if 'LoanAmount' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Loan_Status', y='LoanAmount', data=df, palette='pastel')
    plt.title('Loan Amount Distribution by Loan Status')
    plt.xlabel('Loan Status')
    plt.ylabel('Loan Amount')
    plt.ylim(0, df['LoanAmount'].quantile(0.99)) # Limit y-axis to avoid extreme outliers
    plt.show()

print("\n-- Income Analysis (Applicant + Coapplicant) ---")
if 'ApplicantIncome' in df.columns and 'CoapplicantIncome' in df.columns and 'Loan_Status' in df.columns and not loan_status_missing:
    df['Total_Income'] = df['ApplicantIncome'] + df['CoapplicantIncome']
    plt.figure(figsize=(10, 6))
    sns.boxplot(x='Loan_Status', y='Total_Income', data=df, palette='coolwarm')
    plt.title('Total Income Distribution by Loan Status')
    plt.xlabel('Loan Status')
    plt.ylabel('Total Income')
    plt.ylim(0, df['Total_Income'].quantile(0.99))
    plt.show()
elif 'ApplicantIncome' in df.columns and 'CoapplicantIncome' in df.columns:
    print("\n--- 3.4 Income Analysis (Applicant + Coapplicant) ---")
    print("Cannot show Total Income Distribution by Loan Status as the Loan_Status column was missing or placeholder data was used.")
    df['Total_Income'] = df['ApplicantIncome'] + df['CoapplicantIncome']
    plt.figure(figsize=(10, 6))
    sns.histplot(df['Total_Income'], kde=True, palette='coolwarm')
    plt.title('Total Income Distribution')
    plt.xlabel('Total Income')
    plt.ylabel('Count')
    plt.show()


print("\n--- Descriptive Analytics Complete ---")
if loan_status_missing:
  print("Note: Loan Status analysis is based on placeholder data or was not possible due to the missing 'Loan_Status' column in the original dataset.")
else:
  print("These visualizations and statistics provide insights into the characteristics of approved and rejected loans in your historical data.")
  print("For actual prediction, you would need to build a machine learning model.")
```
