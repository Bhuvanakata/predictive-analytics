```python
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
import os

# Load purchase history dataset
# The UnicodeDecodeError suggests an issue with the file encoding.
# Trying to read with a different encoding like 'latin1'.
try:
    purchase_df = pd.read_csv('shopping dataset.csv', encoding='utf-8')
except UnicodeDecodeError:
    purchase_df = pd.read_csv('shopping dataset.csv', encoding='latin1')

# Add a 'purchased' column with a value of 1 for each transaction to represent a purchase
purchase_df['purchased'] = 1

# Expected columns: user_id, product_id, purchased (1/0)
# Assuming 'CustomerID' is the user_id and 'StockCode' is the product_id and 'Description' is the product_name
purchase_df = purchase_df.rename(columns={'CustomerID': 'user_id', 'StockCode': 'product_id', 'Description': 'product_name'})

# Drop rows with missing user_id
purchase_df.dropna(subset=['user_id'], inplace=True)

# Convert user_id to integer
purchase_df['user_id'] = purchase_df['user_id'].astype(int)

# Create a product ID to product name mapping dictionary
product_id_to_name = purchase_df.set_index('product_id')['product_name'].to_dict()

# Prepare user-item purchase matrix
user_item_matrix = purchase_df.pivot_table(index='user_id', columns='product_id', values='purchased', fill_value=0)

# The rest of the code is the same as your original submission.
def recommend_products(target_user, user_item_matrix, top_n_similar=5, top_n_products=5):
    # Calculate cosine similarity matrix for users
    sim_matrix = cosine_similarity(user_item_matrix)
    sim_df = pd.DataFrame(sim_matrix, index=user_item_matrix.index, columns=user_item_matrix.index)

    # If user interests are not used, consider all other users for similarity
    users_for_similarity = [user for user in user_item_matrix.index if user != target_user]
    if not users_for_similarity:
        return []

    # Get similarity scores for those users
    sim_scores = sim_df.loc[target_user, users_for_similarity]

    # Select top similar users
    top_sim_users = sim_scores.sort_values(ascending=False).head(top_n_similar).index

    # Aggregate their purchases
    similar_users_purchases = user_item_matrix.loc[top_sim_users]
    product_scores = similar_users_purchases.sum(axis=0)

    # Remove products already bought by target user
    target_purchases = user_item_matrix.loc[target_user]
    products_bought = target_purchases[target_purchases > 0].index
    product_scores = product_scores.drop(products_bought, errors='ignore')

    # Recommend top products
    recommended_product_ids = product_scores.sort_values(ascending=False).head(top_n_products).index.tolist()

    return recommended_product_ids

# Example usage:
target_user = 17850  # Replace with actual user_id in your dataset
recommendations = recommend_products(target_user, user_item_matrix)

# Get the product names for the recommended product IDs
recommended_product_names = [product_id_to_name.get(p_id, 'Unknown Product') for p_id in recommendations]

print(f"Recommended products for user {target_user}:")
for name in recommended_product_names:
    print(f"- {name}")
```
